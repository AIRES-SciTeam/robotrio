```python
def _valid_linear_velocity(lin_vel : tuple[float | int | None, ...], relative : bool) -> tuple[bool, str]:
    if not isinstance(lin_vel, tuple) or len(lin_vel) != 2:
        return False, "lin_vel must be of size 2"
    if not (isinstance(lin_vel[0], float | int | type(None)) and isinstance(lin_vel[1], float | int | type(None))):
        return False, "lin_vel args must be float, int or None"
    if not relative:
        return (
            (abs(lin_vel[0] if lin_vel[0] is not None else 0) <= 25) 
            and (abs(lin_vel[1] if lin_vel[1] is not None else 0) <= 18)
        ), "|lin_vel[0]| (X) must be <= 25 and |lin_vel[1]| (Y) must be <= 18"
    return (
        all(-100 <= (lin_vel[i] if lin_vel[i] is not None else 0) <= 100 for i in [0,1]), 
        "lin_vel args must be percent (-100...0...100) in relative mode"
    )
```

```python
_valid_linear_velocity((25, 101), True)
```

```python
def _valid_angular_velocity(ang_vel : float | int | None, relative : bool) -> tuple[bool, str]:
    if not isinstance(ang_vel, float | int | type(None)):
        return False, "ang_vel must be float, int or None"
    return (abs(ang_vel if ang_vel is not None else 0) <= 100), "|ang_vel| must be <= 100"
```

```python
_valid_angular_velocity(101, True)
```
